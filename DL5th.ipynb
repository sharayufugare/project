```python

# Implement Continuous Bag of Words (CBOW) Model

import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import skipgrams
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Lambda, Dense
from tensorflow.keras.preprocessing.sequence import pad_sequences
import tensorflow as tf

# Sample text dataset
text = """
deep learning is a part of machine learning
machine learning helps computers learn
deep learning uses neural networks
"""

# Tokenize the text
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
word_index = tokenizer.word_index
vocab_size = len(word_index) + 1

print("Vocabulary:", word_index)
print("Vocabulary Size:", vocab_size)

# Prepare training data for CBOW
window_size = 2
sequences = tokenizer.texts_to_sequences([text])[0]

X, y = [], []

for i in range(window_size, len(sequences) - window_size):
    context = (
        sequences[i-window_size:i] +
        sequences[i+1:i+window_size+1]
    )
    target = sequences[i]

    X.append(context)
    y.append(target)

X = np.array(X)
y = np.array(y)

print("\nContext words shape:", X.shape)
print("Target words shape:", y.shape)

# Build the CBOW model
embedding_dim = 50

model = Sequential([
    Embedding(input_dim=vocab_size,
              output_dim=embedding_dim,
              input_length=2 * window_size),
    Lambda(lambda x: tf.reduce_mean(x, axis=1)),
    Dense(vocab_size, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train the model
model.fit(X, y, epochs=100, verbose=1)

# Test the model
prediction = model.predict(X[:1], verbose=0)
predicted_id = np.argmax(prediction[0])

index_word = {v: k for k, v in word_index.items()}

print("\nContext words:")
print([index_word[i] for i in X[0]])

print("Actual target word:", index_word[y[0]])
print("Predicted target word:", index_word[predicted_id])
```
